##Filtrando mais valores do DataFrame

```python
import pandas as pd
data = pd.read_csv('https://raw.githubusercontent.com/Laisgraca/Pandas/refs/heads/main/gasprices.csv', sep='\t')
data
```

##**Selecionando registros do posto do Amazonas e com preços acima de 2 Reais**






```python
data['ESTADO'].unique()
```

```python
data['ESTADO'] == 'AMAZONAS'
```

```python
data['PREÇO MÉDIO REVENDA'] > 2.0
```

```python
(data['ESTADO'] =='AMAZONAS') & (data['PREÇO MÉDIO REVENDA'] > 2.0)
```

Note que o resultado da seleção continua sendo uma _Series de booleans_ com o mesmo número de linhas do Dataframe, de modo que cada linha possuirá
um valor booleano indicando se o posto é do Amazonas **E** o preço aferido do combustível é maior que 2 reais (True) ou não.
O simbolo & significa **AND** na comparação. Essa nomeclatura do python/pandas é diferente das nomeclaturas tradicionais (&&). similarmente:


*   **I** representa **OR** (Não é II)
*   **~** representa **NOT** (Não é !)




```python
selecao = (data['ESTADO'] =='AMAZONAS') & (data['PREÇO MÉDIO REVENDA'] > 2.0)
data[selecao]
```

Alternativamente ,poderíamos usar o método query para fazermos tal seleção. Porém, isso não é possível especificamente para esse caso , pois o rótulo da coluna 'PREÇO MÉDIO REVENDA' possui caracteres inválidos para o método(cedilha,acentos)





```python
#Não funciona
data.query('ESTADO == "AMAZONAS" and PREÇO MÉDIO REVENDA > 2')
```

```python
#Funciona colocando (`) crase para caracteres inválidas
data.query('ESTADO == "AMAZONAS" and `PREÇO MÉDIO REVENDA`>2')
```

```python
#Funciona pois não há caracteres inválidos
data.query('ESTADO == "AMAZONAS" and PRODUTO == "GASOLINA COMUM"')
```

```python
(data['ESTADO'] == 'AMAZONAS') | (data['ESTADO'] == 'SAO PAULO') # retorna o valor em boolean
selecao = (data['ESTADO'] == 'AMAZONAS') | (data['ESTADO'] == 'SAO PAULO')
data[selecao]# retorna a lista
```

```python
#Método query utilizamos (o or , and e not)
data.query('ESTADO == "AMAZONAS" or  ESTADO == "SAO PAULO"')
```
