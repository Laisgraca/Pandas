#**Selecionando registros Região,Estado e Produto.**

```python
import pandas as pd
```

```python
data = pd.read_csv('https://raw.githubusercontent.com/Laisgraca/Pandas/refs/heads/main/gasprices.csv', sep='\t')
data.head()
```

```python
selecao = (data['PRODUTO'] == 'GASOLINA COMUM') & (data['ESTADO'] == 'SAO PAULO') & (data['REGIÃO'] =='SUDESTE')
data[selecao]
```

```python
data[selecao][['PRODUTO','ESTADO','REGIÃO']]
```

##**Alternativa 2**

```python
lista_regiao = ['NORTE','NORDESTE','SUDESTE']
selecao = data['REGIÃO'].isin(lista_regiao) # Pergunta se existe algum valor contido na lista igual a da Linha
```

##**Alternativa 3**

```python
lista_regiao = ['NORTE','NORDESTE','SUDESTE']
```

```python
# Utiliza o @ para comparar com uma variável
data.query('`REGIÃO` in @lista_regiao')
```

##**Iterando  com DataFrames**
