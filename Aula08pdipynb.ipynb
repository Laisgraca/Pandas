# Label-based selection (Seleção baseadas em rótulos)
Método loc: seleciona elementos do Dataframe , baseado em seus rótulos ---> row-first,column-second

```python
import pandas as pd

```

```python
pesquisa_satisfacao = pd.DataFrame({
    'Bom': [50,70,100],
    'Ruim': [20,40,60],
    'Péssimo':[10,5,3]
  },index=['XboxOne','Playstation4', 'Switch'])
print(pesquisa_satisfacao)
```

```python
# Retorna a linha de índice 0 (implícito) ----> usando iloc
pesquisa_satisfacao.iloc[0]
```

```python
# Retorna o valor da linha de índice 0 e coluna de índice 1 ---> usando iloc
pesquisa_satisfacao.iloc[0, 1]
```

```python
# Retorna a linha cujo o rótulo do índice é 'XboxOne'
pesquisa_satisfacao.loc['XboxOne']
```

```python
# Não funciona ----> iloc tentando acessar índices com rótulos
pesquisa_satisfacao.iloc['XboxOne']
```

```python
# Não funciona ----> loc tentando acessar índices rotulados com números
pesquisa_satisfacao.loc[0]
```

```python
pesquisa_satisfacao
```

```python
# Retorna o valor da linha 'Switch' , coluna 'ruim'
pesquisa_satisfacao.loc['Switch','Ruim']
```

```python
# Selecionando as linha de índices rotulados 'XboxOne', 'Switch
pesquisa_satisfacao.loc[['XboxOne', 'Switch']]
```

```python
# Retorne todas as linhas e apenas as colunas com rótulos 'Bom' e 'Péssimo'
pesquisa_satisfacao[['Bom','Péssimo']]

```

```python
# Retorna todas as linhas e apenas as colunas com rótulos 'Bom' e 'Péssimo'
pesquisa_satisfacao.loc[: ,['Bom','Péssimo']]
```

```python
pesquisa_satisfacao.loc['XboxOne': 'Switch']
```
