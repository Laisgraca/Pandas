#Seleção de Índices

```python
import pandas as pd
data = pd.read_csv('https://raw.githubusercontent.com/Laisgraca/Pandas/refs/heads/main/gasprices.csv', sep='\t')
```

# Index-based selection (Seleção baseada em índices)

```python
# Mostrando linhas específicas de um Dataframe
#iloc: seleciona elemento do DataFrame baseado em seu índice(número)row-first,column-second

data.iloc[1]
```

# Selecionando Múltiplas observações/linhas

```python
# Selecionando as linhas de índice 0 a 5(incluso)
data.iloc[0:6]
```

```python
#Selecionando as linhas de 10 a 15(incluso)
data.iloc[10:16]
```

```python
# Selecionando as linhas/observações de índice 1,5,10,15 (basta passar uma lista)
data.iloc[[1,5,10,15]]
```

```python
# Selecionando as linhas/observações de índice 10,5,1,15 (basta passar uma lsta)
data.iloc[[10,5,1,15]]
```

```python
# Retornar o valor da linha de índice 2, coluna 3 ('ESTADO')
data.iloc[2,3]
```
