##**Alternativa 2**: Utilizando o método query(pergunta)

```python
data.query('ESTADO == "SAO PAULO" ')
```

Uma boa prática é **salvar o Dataframe filtrado em uma nova variável**.  Isso simplifica a complexidade do código para futuras análises.

```python
postos_sp = data.query('ESTADO == "SAO PAULO"')

```

##**Resetando índices**
Note que os índices das linhas/registros após a filtragem continuam sendo os mesmos do Dataframe original.
Em muitas situações, manter esta informação será importante.
mas se você quiser **resetar** os índices do dataframe, fazendo com que os registros começem com o índice até num_linhas-1 ,use o método: reset_index.

```python
postos_sp.head()
```

```python
#Veja que ele criou um novo índice e ao mesmo tempo manteve o original
postos_sp.reset_index()
```

```python
#Se desejar retirar o índice original utilizamos o drop
postos_sp.reset_index(drop=True, )
```

```python
#Note que postos_sp ainda não foi alterado ,
#precisamos utilizar o método inplace = True
postos_sp
```

```python
#Utilizando o método inplace = True e
#fazendo as devidas alterações eliminando índice
postos_sp.reset_index(drop = True, inplace = True)
```

```python
postos_sp
```

```python
#ALTERNATIVA MAIS COMUM
#Não precisa passar o inplace = True porque ele gera uma cópia dos dados filtrados
# e em cima dessa cópia estamos gerando alterações que vai dar origem a outra cópia
postos_sp = data.query('ESTADO == "SAO PAULO"').reset_index(drop=True)
```

```python
postos_sp
```

#**Selecionando Registros de postos do Estado de São Paulo ou do Rio de Janeiro com Gasolina Comum acima de 2 reais**

```python
postos = data.query('ESTADO == "SAO PAULO" or ESTADO == "RIO DE JANEIRO"')
posto_2 = postos.query("`PREÇO MÉDIO REVENDA` > 2.0  and PRODUTO == 'GASOLINA COMUM'")
posto_2

```

```python
#Em uma linha só
posto_2 = data.query('ESTADO in ["SAO PAULO", "RIO DE JANEIRO"] and `PREÇO MÉDIO REVENDA` > 2.0')

posto_2
```
