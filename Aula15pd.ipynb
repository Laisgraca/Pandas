#**Selecionando registros de postos de São Paulo ou Rio de Janeiro com Gasolina Comum acima de 2 reais.**
Podemos fazer a filtragem com múltiplos condicionais em partes:

```python
import pandas as pd
data = pd.read_csv('https://raw.githubusercontent.com/Laisgraca/Pandas/refs/heads/main/gasprices.csv', sep='\t')
data.head()
```

```python
data['ESTADO'].unique()
```

```python
#filtragem selecionando os valores de SP ou RJ
selecao_1= (data['ESTADO']=='RIO DE JANEIRO' )| (data['ESTADO']=='SAO PAULO')
selecao_2 = (data['PRODUTO'] == 'GASOLINA COMUM')
selecao_3= (data['PREÇO MÉDIO REVENDA'] > 2)
selecao_final = selecao_1 & selecao_2 & selecao_3

```

```python
#Filtragem final
data_filtrado = data[selecao_final]
data_filtrado
```

```python
data_filtrado['ESTADO'].unique()
data_filtrado['PRODUTO'].unique()
```

```python
#Primeira filtragem usando o método query
selecao_1 = data.query('ESTADO == "RIO DE JANEIRO" or ESTADO == "SAO PAULO"')
selecao_1
```

```python
#Segunda Filtragem usando o método query
selecao_2 = selecao_1.query('PRODUTO == "GASOLINA COMUM" and `PREÇO MÉDIO REVENDA` > 2.0')
selecao_2
```

```python
#Se quiser só selecionar a coluna produto
selecao_2['PRODUTO']
```

```python
#Se quiser selecionar mais colunas
selecao_2[['ESTADO','PRODUTO','PREÇO MÉDIO REVENDA']]
```
